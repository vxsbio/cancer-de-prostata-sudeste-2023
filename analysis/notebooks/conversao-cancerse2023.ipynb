```python
import pyreaddbc
import sqlite3
import pandas as pd
from dbfread import DBF
```

# Conversão dos arquivos de dados utilizados  

### 1. Convertendo os arquivos .dbf baixados do DATASUS para .db
Nessa etapa é realizado a conversão dos arquivos baixados do DATASUS '.dbc' para o formato '.db'.  
Note que todos esses dados são do SIM (Sistema de Informações de Mortalidade), com o tipo de arquivo sendo DO (Declaração de Óbito)  
Dados: https://datasus.saude.gov.br/transferencia-de-arquivos#

```python
conexao = sqlite3.connect("cancer_sudeste2023.db")

ficheiro_dbc = (
    r"E:\Meus arquivos\Projects\Cancer\dados-sus-SIM\DOSP2023.dbc",
    r"E:\Meus arquivos\Projects\Cancer\dados-sus-SIM\DORJ2023.dbc",
    r"E:\Meus arquivos\Projects\Cancer\dados-sus-SIM\DOMG2023.dbc",
    r"E:\Meus arquivos\Projects\Cancer\dados-sus-SIM\DOES2023.dbc"
    )
```

```python
for arquivo_dbc in ficheiro_dbc:
    arquivo_dbf = arquivo_dbc.replace(".dbc", ".dbf")
    dbf_data = pyreaddbc.dbc2dbf(arquivo_dbc, arquivo_dbf)
    df_sim = pd.DataFrame(iter(DBF(arquivo_dbf)))
    df_sim.to_sql(
        "mortalidade",
        conexao,
        if_exists="append",
        index=False
    )
conexao.close()
```

### 2. Término da conversão
Após o término da conversão, o novo arquivo final no formato .db "cancer_sudeste2023.db" é gerado na pasta atual do projeto pronto para pré-processamento em SQL.
